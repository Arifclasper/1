## Part A — Setup
- Report dataset size (rows, columns)

```python
import pandas as pd
import numpy as np

print("Libraries pandas and numpy imported successfully.")
```

```python
df_wine = pd.read_csv('wine_dataset.csv')
print("Wine dataset size:", df_wine.shape)
print(df_wine.head())
```

## Part B — Simple Random Sampling

```python
sample_size = 50
srs_wine = df_wine.sample(n=sample_size, random_state=42)
print("Wine Population mean (alcohol):", df_wine['alcohol'].mean())
print("Wine Sample mean (alcohol):", srs_wine['alcohol'].mean())
```

## Part C — Systematic Sampling

```python
n = 50
k = len(df_wine) // n
start = np.random.randint(0, k)
sys_wine = df_wine.iloc[start::k][:n]
print("Wine Systematic sample mean (alcohol):", sys_wine['alcohol'].mean())
```

## Part D — Stratified Sampling

```python
strata_col = "quality"  # categorical column
frac = sample_size / len(df_wine)
strat_wine = df_wine.groupby(strata_col, group_keys=False).sample(frac=frac, random_state=42)
print("Wine Stratified sample mean (alcohol):", strat_wine['alcohol'].mean())
```

## Part E — Cluster Sampling

```python
df_wine['cluster_id'] = df_wine.index // (len(df_wine)//10)  # 10 clusters
clusters_wine = np.random.choice(df_wine['cluster_id'].unique(), size=2, replace=False)
cluster_wine = df_wine[df_wine['cluster_id'].isin(clusters_wine)]
print("Wine Selected clusters:", clusters_wine)
print("Wine Cluster sample mean (alcohol):", cluster_wine['alcohol'].mean())
```

## Part F — Comparison & Reflection
Compare sample means vs population mean, then write your reflection.

```python
print("\n--- Wine Quality Comparison ---")
print("Population mean (alcohol):", df_wine['alcohol'].mean())
print("SRS mean:", srs_wine['alcohol'].mean())
print("Systematic mean:", sys_wine['alcohol'].mean())
print("Stratified mean:", strat_wine['alcohol'].mean())
print("Cluster mean:", cluster_wine['alcohol'].mean())
```
