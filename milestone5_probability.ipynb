# Milestone 5 – Probability
This notebook contains sample code for each section of Milestone 5.

## Section A: Setup & Load Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Load your dataset
# Replace with the actual filename
# If the file is not in your Colab environment, uncomment the line below to upload it:
# uploaded = files.upload()
# Then ensure the filename in pd.read_csv matches the uploaded file's name.
df = pd.read_csv('wine_dataset.csv')   # replace with your actual file path
df.head()
```

## Section B: Treat Dataset as Sample Space

```python
N = len(df)
print('Total observations:', N)
```

## Section C: Task 1 – Defining Events

```python
# Example event definitions (change column names as needed)
A = df[df['alcohol'] > 11]                     # Event A: Alcohol > 11%
B = df[df['quality'] >= 7]                     # Event B: High-quality wine (quality ≥ 7)
C = df[df['residual_sugar'] > 5]               # Event C: Residual sugar > 5 g/L

print('Event A size:', len(A))
print('Event B size:', len(B))
print('Event C size:', len(C))
```

## Section D: Task 2 – Calculating Basic Probability

```python
P_A = len(A) / N
P_B = len(B) / N
P_C = len(C) / N

print('P(A) =', P_A)
print('P(B) =', P_B)
print('P(C) =', P_C)
```

## Section E: Task 3 – Combined Events

```python
A_int_B = df[(df['alcohol'] > 11) & (df['quality'] >= 7)]
P_A_int_B = len(A_int_B) / N
print('P(A ∩ B) =', P_A_int_B)

A_union_B = df[(df['alcohol'] > 11) | (df['quality'] >= 7)]
P_A_union_B = len(A_union_B) / N
print('P(A ∪ B) =', P_A_union_B)

A_comp = df[df['alcohol'] <= 11]
P_A_comp = len(A_comp) / N
print('P(Aᶜ) =', P_A_comp)

rule_value = P_A + P_B - P_A_int_B
print('Rule Verification:', rule_value)
print('Actual P(A ∪ B):', P_A_union_B)
```

## Section F: Visualization

```python
counts = [len(A), len(A_comp)]
labels = ['A: Alcohol > 11%', 'Aᶜ: Alcohol ≤ 11%']

plt.bar(labels, counts, color=['skyblue','lightcoral'])
plt.title('Event A and Complement Frequency')
plt.ylabel('Count')
plt.show()
```

```python
df['quality'].value_counts().sort_index().plot(kind='bar', color='seagreen')
plt.title('Wine Quality Frequency')
plt.ylabel('Count')
plt.xlabel('Quality Score')
plt.show()
```

## Section G: Summary and Reflection:

# Event C (Residual sugar > 5) captures wines with sweeter profiles.
# The union and intersection rules verify correctly, showing consistency in probability laws.
