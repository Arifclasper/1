This notebook contains tasks to calculate mean, median, mode, variance, and standard deviation for selected numerical columns of your dataset.

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
```

```python
# Load your dataset
df = pd.read_csv('wine_dataset.csv')  # Replace with your dataset path
df.head()
```

## Task 1: Measures of Central Tendency
Select at least two numerical columns and calculate mean, median, and mode.

```python
# Select numerical columns: Alcohol and Residual Sugar
num_cols = ['alcohol', 'residual_sugar']

for col in num_cols:
    print(f'\nColumn: {col}')
    mean_val = df[col].mean()
    median_val = df[col].median()
    mode_val = df[col].mode().tolist()
    print(f'Mean: {mean_val:.2f}')
    print(f'Median: {median_val:.2f}')
    print(f'Mode: {mode_val}')
```

## Task 2: Measures of Dispersion
Calculate variance and standard deviation for the selected columns.

```python
for col in num_cols:
    print(f'\nColumn: {col}')
    variance_val = df[col].var()
    std_dev_val = df[col].std()
    print(f'Variance: {variance_val:.2f}')
    print(f'Standard Deviation: {std_dev_val:.2f}')
```

## Task 3: Visualization
Plot histograms with mean, median, and mode indicated.

```python
for col in num_cols:
    plt.figure(figsize=(8,5))
    sns.histplot(df[col], bins=10, kde=False, color='skyblue', edgecolor='black')
    plt.axvline(df[col].mean(), color='red', linestyle='dashed', linewidth=1.5, label='Mean')
    plt.axvline(df[col].median(), color='green', linestyle='dashed', linewidth=1.5, label='Median')
    plt.axvline(df[col].mode()[0], color='orange', linestyle='dashed', linewidth=1.5, label='Mode')
    plt.title(f'Histogram of {col} (Wine Quality)')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.legend()
    plt.show()
```

## Task 4: Analysis and Conclusion
Alcohol: Mean and median are close, showing a fairly balanced distribution. Slight right‑skew due to a few higher‑alcohol wines.

Residual Sugar: More skewed, with most wines having low sugar but some outliers with very high values.

Dispersion: Residual sugar has higher variance and standard deviation than alcohol, meaning sugar levels vary more widely across wines.
